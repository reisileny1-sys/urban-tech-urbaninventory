Notebook para explorar os conjuntos e relações do domínio. Rode a partir da pasta `notebooks/`.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

from src.data.manuseio_dados import carregar_csv
from src.algorithms import conjuntos as cj

inv = carregar_csv(os.path.join('..', 'data', 'raw'))
print(len(inv.ativos), 'ativos')
```

## Conjuntos A = Norte e B = Pendentes

```python
A = inv.ativos_da_jurisdicao('Norte')
B = inv.ativos_com_manutencao_pendente()
print('A     =', sorted(A))
print('B     =', sorted(B))
print('A ∩ B =', sorted(cj.intersecao(A, B)))
print('A − B =', sorted(cj.diferenca(A, B)))
print('A ∪ B =', sorted(cj.uniao(A, B)))
```

## Jurisdições são disjuntas dois a dois?

```python
jur = list(inv.jurisdicoes)
for i in range(len(jur)):
    for j in range(i + 1, len(jur)):
        a, b = inv.ativos_da_jurisdicao(jur[i]), inv.ativos_da_jurisdicao(jur[j])
        print(jur[i], '∩', jur[j], '=', cj.intersecao(a, b) or '∅')
```

## Relações e composição

```python
R1 = inv.relacao_usuario_orgao()
R2 = inv.relacao_orgao_ativo()
print('R1 =', sorted(R1))
for u in inv.usuarios:
    print(u, '→', sorted(cj.imagem(cj.composicao_indexada(R1, R2), u)))
```
